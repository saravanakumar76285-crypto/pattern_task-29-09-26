***PATTERN TASK***

Q.print the below pattern    

enter the number between 0-9:9  
1  
22  
333  
4444  
55555  
666666  
7777777  
88888888  
999999999  

```python
num=int(input("enter the number between 0-9:"))
for i in range(1,num+1):
    for j in range(1,i+1):
        print(i,end="")
    print()
```
